```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
order_items_df = pd.read_csv("D:\Projects\Cart2Insights\Data\Cleaned\clist_order_items_dataset.csv")

products_df =pd.read_csv("D:\Projects\Cart2Insights\Data\Cleaned\clist_products_dataset.csv")
```

```python
merged_df = pd.merge(order_items_df,products_df, on="product_id", how="inner")
```

```python
#plt.figure(figsize=(10, 6))
sns.regplot(
    data=merged_df, 
    x='product_weight_g', 
    y='price', 
    scatter_kws={'alpha': 0.3, 'color': 'green'}, 
    line_kws={'color': 'red'}
)

plt.title('Bivariate Analysis: Product Weight vs. Price')
plt.xlabel('Product Weight (grams)')
plt.ylabel('Price ($)')
plt.grid(True, linestyle='--')
plt.show()
```
